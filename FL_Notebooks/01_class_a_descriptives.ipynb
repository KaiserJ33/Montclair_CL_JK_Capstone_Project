```python
import pandas as pd
import numpy as np

crimuser["Age"] = pd.to_numeric(crimuser["Age"], errors="coerce")

crimuser["Length"] = crimuser["Username"].str.len()

crimuser["DigitCount"] = crimuser["Username"].str.count(r"\d")

crimuser["LetterCount"] = crimuser["Username"].str.count(r"[A-Za-z]")

crimuser["VowelCount"] = crimuser["Username"].str.count(r"[AEIOUaeiou]")

print("="*60)
print("DATASET SUMMARY")
print("="*60)

print(f"Total Records: {len(crimuser)}")
print(f"Unique Usernames: {crimuser['Username'].nunique()}")
print(f"Duplicate Usernames: {crimuser['Username'].duplicated().sum()}")

print("\n")
print("="*60)
print("USERNAME LENGTH")
print("="*60)

print(crimuser["Length"].describe())

print("\nMean:", crimuser["Length"].mean())
print("Median:", crimuser["Length"].median())
print("Standard Deviation:", crimuser["Length"].std())
print("Variance:", crimuser["Length"].var())
print("Minimum:", crimuser["Length"].min())
print("Maximum:", crimuser["Length"].max())


print("\n")
print("="*60)
print("AGE STATISTICS")
print("="*60)

print(crimuser["Age"].describe())

print("\nMean Age:", crimuser["Age"].mean())
print("Median Age:", crimuser["Age"].median())
print("Standard Deviation:", crimuser["Age"].std())

print("\n")
print("="*60)
print("SEX DISTRIBUTION")
print("="*60)

print(crimuser["Sex"].value_counts())

print("\nPercentages")

print((crimuser["Sex"].value_counts(normalize=True)*100).round(2))

print("\n")
print("="*60)
print("NATIONALITY")
print("="*60)

print(crimuser["Nationality"].value_counts())

print("\nPercentages")

print((crimuser["Nationality"].value_counts(normalize=True)*100).round(2))

print("\n")
print("="*60)
print("CRIME TYPES")
print("="*60)

print(crimuser["Crime"].value_counts())

print("\nPercentages")

print((crimuser["Crime"].value_counts(normalize=True)*100).round(2))

print("\n")
print("="*60)
print("USERNAME FEATURES")
print("="*60)

print(f"Average Digit Count: {crimuser['DigitCount'].mean():.2f}")

print(f"Average Letter Count: {crimuser['LetterCount'].mean():.2f}")

print(f"Average Vowel Count: {crimuser['VowelCount'].mean():.2f}")

print(f"Contains Digit (%): {(crimuser['Username'].str.contains(r'\\d')).mean()*100:.2f}")

print(f"Ends with Digit (%): {(crimuser['Username'].str.contains(r'\\d$')).mean()*100:.2f}")

print(f"Starts with Letter (%): {(crimuser['Username'].str.match(r'^[A-Za-z]')).mean()*100:.2f}")

print(f"Contains Underscore (%): {(crimuser['Username'].str.contains('_')).mean()*100:.2f}")

print(f"Contains Period (%): {(crimuser['Username'].str.contains(r'\\.')).mean()*100:.2f}")

print(f"Contains Other Special Character (%): {(crimuser['Username'].str.contains(r'[^A-Za-z0-9_.]')).mean()*100:.2f}")

print(f"All Lowercase (%): {(crimuser['Username'].str.fullmatch(r'[a-z0-9_.]+')).mean()*100:.2f}")

print(f"All Uppercase (%): {(crimuser['Username'].str.fullmatch(r'[A-Z0-9_.]+')).mean()*100:.2f}")

print(f"Mixed Case (%): (((crimuser['Username'].str.contains('[A-Z]')) & (crimuser['Username'].str.contains('[a-z]'))).mean()*100):.2f")

print(f"Repeated Characters 3+ (%): {(crimuser['Username'].str.contains(r'(.)\\1{2,}', regex=True)).mean()*100:.2f}")

print("\n")
print("="*60)
print("USERNAME LENGTH BY SEX")
print("="*60)

print(
    crimuser.groupby("Sex")["Length"].agg([
        "count",
        "mean",
        "median",
        "std",
        "min",
        "max"
    ])
)

print("\n")
print("="*60)
print("AGE BY SEX")
print("="*60)

print(
    crimuser.groupby("Sex")["Age"].agg([
        "count",
        "mean",
        "median",
        "std",
        "min",
        "max"
    ])
)

print("\n")
print("="*60)
print("USERNAME LENGTH BY CRIME TYPE")
print("="*60)

print(
    crimuser.groupby("Crime")["Length"].agg([
        "count",
        "mean",
        "median",
        "std",
        "min",
        "max"
    ])
)


print("\n")
print("="*60)
print("AGE BY CRIME TYPE")
print("="*60)

print(
    crimuser.groupby("Crime")["Age"].agg([
        "count",
        "mean",
        "median",
        "std",
        "min",
        "max"
    ])
)

print("\n")
print("="*60)
print("SEX x CRIME")
print("="*60)

print(pd.crosstab(crimuser["Sex"], crimuser["Crime"]))

print("\n")
print("="*60)
print("SEX x NATIONALITY")
print("="*60)

print(pd.crosstab(crimuser["Sex"], crimuser["Nationality"]))

print("\n")
print("="*60)
print("CORRELATION MATRIX")
print("="*60)

print(
    crimuser[
        ["Age",
         "Length",
         "DigitCount",
         "LetterCount",
         "VowelCount"]
    ].corr()
)

print("\n")
print("="*60)
print("USERNAME LENGTH QUARTILES")
print("="*60)

print(crimuser["Length"].quantile([0.25,0.50,0.75]))

print("\n")
print("="*60)
print("MOST COMMON USERNAMES")
print("="*60)

print(crimuser["Username"].value_counts().head(20))
```
